# Customer Segmentation for Hospitality Marketing

This notebook demonstrates unsupervised learning (KMeans clustering) for segmenting hotel customers based on behavioral features.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
np.random.seed(42)
```

```python
# Simulated dataset
data = pd.DataFrame({
    'age': np.random.randint(20, 70, 500),
    'visits_per_year': np.random.poisson(3, 500),
    'avg_spend': np.random.normal(250, 50, 500),
    'review': np.random.choice(['great service', 'poor experience', 'very clean', 'not worth the price'], 500),
    'sentiment': np.random.choice([0, 1], 500)
})
X = data[['age', 'visits_per_year', 'avg_spend']] if 'review' not in data.columns else None
```

```python
from sklearn.cluster import KMeans
model = KMeans(n_clusters=3).fit(X)
data['segment'] = model.labels_
sns.pairplot(data, hue='segment')
plt.suptitle('Customer Segments', y=1.02)
plt.show()
```
