# Time Series Forecasting for Hotel Bookings

Use time series forecasting to predict hotel bookings over the next 30 days.

```python
import pandas as pd
import numpy as np
np.random.seed(42)
data = pd.DataFrame({
    'age': np.random.randint(20, 70, 500),
    'visits_per_year': np.random.poisson(3, 500),
    'avg_spend': np.random.normal(250, 50, 500)
})
```

```python
import matplotlib.pyplot as plt
from statsmodels.tsa.holtwinters import ExponentialSmoothing
ts = pd.Series(np.random.poisson(100, 180))
model = ExponentialSmoothing(ts, trend='add').fit()
forecast = model.forecast(30)
plt.plot(ts, label='Historical')
plt.plot(range(len(ts), len(ts)+30), forecast, label='Forecast')
plt.legend(); plt.title('Hotel Booking Forecast')
plt.show()
```
