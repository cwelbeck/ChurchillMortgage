# Named Entity Recognition (NER) from Electronic Health Records (EHR) 🏥📑
This notebook extracts clinical entities (like diseases, medications, anatomy) from EHR notes using a biomedical NER model.

```
!pip install transformers gradio --quiet
```

```

from transformers import pipeline
import gradio as gr
```

```

# Load a medical-specific NER pipeline
ner = pipeline("ner", model="d4data/biomedical-ner-all", aggregation_strategy="simple")
```

```

def extract_entities(text):
    results = ner(text)
    return "\n".join([f"{r['entity_group']}: {r['word']}" for r in results])
```

```

gr.Interface(
    fn=extract_entities,
    inputs=gr.Textbox(lines=10, placeholder="Paste EHR note here..."),
    outputs="text",
    title="EHR NER Extractor",
    description="Extract named medical entities (conditions, drugs, anatomy) from EHR notes using a BioNER model."
).launch()
```
