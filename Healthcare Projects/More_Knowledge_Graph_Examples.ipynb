```
# 📚 Import libraries
from rdflib import Graph, URIRef, Literal, Namespace, RDF
import gradio as gr
```

```
# 🌐 Define a simple knowledge graph using RDFLib
EX = Namespace("http://example.org/")

g = Graph()
g.bind("ex", EX)

# Add example triples
g.add((EX.Alice, RDF.type, EX.Person))
g.add((EX.Alice, EX.knows, EX.Bob))
g.add((EX.Bob, RDF.type, EX.Person))
g.add((EX.Bob, EX.knows, EX.Carol))
g.add((EX.Carol, RDF.type, EX.Person))
```

```
# 🔍 Query knowledge graph for people and their connections
def query_people(graph):
    q = '''
    SELECT ?person ?knows WHERE {
        ?person rdf:type ex:Person .
        ?person ex:knows ?knows .
    }
    '''
    return graph.query(q)
```

```
# 🎛️ Interface function to explore relationships
def explore_kg():
    result = query_people(g)
    return "\n".join([f"{row.person.split('/')[-1]} knows {row.knows.split('/')[-1]}" for row in result])
```

```
# 🎨 Gradio UI
with gr.Blocks() as demo:
    gr.Markdown("## 🤝 Simple Knowledge Graph Explorer")
    output = gr.Textbox(label="People Relationships")
    btn = gr.Button("Query Graph")
    btn.click(fn=explore_kg, outputs=output)

demo.launch()
```
