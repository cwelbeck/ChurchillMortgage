# Clinical Note Summarizer 🩺📄
This notebook summarizes free-text clinical notes using a transformer model.

```
!pip install transformers datasets gradio --quiet
```

```

from transformers import pipeline
import gradio as gr
```

```

# Load the summarizer model (fine-tuned for clinical text)
summarizer = pipeline("summarization", model="Falconsai/text_summarization")
```

```

def summarize_notes(note_text):
    summary = summarizer(note_text, max_length=150, min_length=30, do_sample=False)[0]['summary_text']
    return summary
```

```

gr.Interface(
    fn=summarize_notes,
    inputs=gr.Textbox(lines=10, placeholder="Paste your clinical notes here..."),
    outputs="text",
    title="Clinical Note Summarizer",
    description="Summarizes long clinical notes into concise medical summaries using a transformer model."
).launch()
```
