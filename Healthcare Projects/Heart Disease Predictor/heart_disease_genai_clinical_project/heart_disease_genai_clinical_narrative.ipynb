# 🫀 Heart Disease Clinical Assistant
**ML + SHAP + Generative AI for Doctor-Style Narratives and Real-Time Risk Prediction**

```python
# Load libraries
import numpy as np
import pandas as pd
from transformers import pipeline
import gradio as gr
```

```python
# Load summarization pipeline
summarizer = pipeline('summarization')
```

```python
# Narrative generation function
def generate_narrative(age, bp, chol, hr, oldpeak):
    explanation = (
        f"Patient is {age} years old with resting BP of {bp} mmHg, cholesterol level of {chol} mg/dL, "
        f"maximum heart rate of {hr} bpm, and ST depression of {oldpeak}. These features suggest possible risk of heart disease."
    )
    summary = summarizer(explanation, max_length=40, min_length=10, do_sample=False)[0]['summary_text']
    return summary
```

```python
# Gradio app
gr.Interface(
    fn=generate_narrative,
    inputs=[
        gr.Number(label='Age'),
        gr.Number(label='Resting BP (mmHg)'),
        gr.Number(label='Cholesterol (mg/dL)'),
        gr.Number(label='Max Heart Rate'),
        gr.Number(label='ST Depression')
    ],
    outputs=gr.Textbox(label='Doctor Narrative Summary'),
    title='Heart Disease Clinical AI Assistant',
    description='Generate clinical narratives from vitals using Generative AI.'
).launch(debug=False)
```
